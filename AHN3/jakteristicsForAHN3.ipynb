```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import os
import warnings

warnings.filterwarnings("ignore")
```

```python
def read_csv_point_cloud(csv_file):
    data = pd.read_csv(csv_file, header=None)
    return data
```

```python
fileName = "69EN2" #"32CN2_11"
```

```python
csv_file = fr"F:\Aditya\Tiles\AHN3 Tiles\Raw Tiles\{fileName}.csv"
points = read_csv_point_cloud(csv_file)
points
```

```python

points = points[[0, 1, 2, 10]].copy()
points.columns = ['x', 'y', 'z', 'label']

points = points.iloc[1:].copy()

points['label'] = points['label'].astype(float).astype(int)

cols_to_float = ['x', 'y', 'z']
points[cols_to_float] = points[cols_to_float].astype(float)
```

```python
UTM_OFFSET = [-195000.00, -319000.00, 0.00]
"""
We need to add these values to x,y,z
"""
cols_to_float = ['x', 'y', 'z']
points[cols_to_float] = points[cols_to_float].astype(float)

points['x'] = points['x'] + UTM_OFFSET[0]
points['y'] = points['y'] + UTM_OFFSET[1]
points['z'] = points['z'] + UTM_OFFSET[2]
```

```python
classes = points['label']
```

```python
points.drop(columns=['label'], inplace=True)
```

```python
xyz = points.to_numpy()
```

```python
xyz
```

```python
search_radius= 0.35 #0.10 #0.15 # 0.25
radText= '035' #'010' #'015'
```

```python
%%time

from jakteristics import las_utils, compute_features, FEATURE_NAMES
print(FEATURE_NAMES)
features = compute_features(xyz, search_radius=search_radius, feature_names=['eigenvalue_sum', 'omnivariance', 'eigenentropy', 'anisotropy', 'planarity', 'linearity', 'PCA1', 'PCA2', 'surface_variation', 'sphericity', 'verticality'])
```

```python
features = pd.DataFrame(features, columns=['eigenvalue_sum', 'omnivariance', 'eigenentropy', 'anisotropy', 'planarity', 'linearity', 'PCA1', 'PCA2', 'surface_variation', 'sphericity', 'verticality'])
```

```python
points_with_features = pd.concat([points, features], axis=1)
points_with_features = pd.concat([points_with_features, classes], axis=1)
points_with_features
```

```python
%%time

points_with_features.to_csv(f'F:\Aditya\Tiles\AHN3 Tiles\Processed_Tiles\{radText}_{fileName}.csv')
print("DONE!")
```
